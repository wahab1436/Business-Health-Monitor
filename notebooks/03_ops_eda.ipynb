# Operations EDA — Business Health Monitor

SLA trends, incident pattern analysis, uptime and resolution time exploration.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
from data.generators.ops_generator import generate_ops_data

df = generate_ops_data(n_days=730, seed=42)
df['date'] = pd.to_datetime(df['date'])
print(df.shape)
df.head()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

df.plot(x='date', y='sla_compliance_pct', ax=axes[0,0], title='SLA Compliance (%)', color='blue')
axes[0,0].axhline(95, color='green', linestyle='--', label='Target 95%')

df.plot(x='date', y='system_uptime_pct', ax=axes[0,1], title='System Uptime (%)', color='green')

df.plot(x='date', y='avg_resolution_hours', ax=axes[1,0], title='Avg Resolution Time (h)', color='orange')

df.plot.scatter(x='ticket_volume', y='avg_resolution_hours',
    ax=axes[1,1], title='Tickets vs Resolution', alpha=0.3,
    c=df['anomaly_flag'].map({True:'red', False:'steelblue'}))

plt.tight_layout(); plt.show()
```

```python
print('Ops summary stats:')
print(df[['sla_compliance_pct','system_uptime_pct','avg_resolution_hours','ticket_volume']].describe().round(2))
print(f'\nIncident days: {df["anomaly_flag"].sum()} ({df["anomaly_flag"].mean()*100:.1f}%)')
```
