# Model Evaluation — Business Health Monitor

Full evaluation of all 5 ML models: confusion matrices, ROC curves, feature importance.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import yaml

with open('../config.yaml') as f:
    CONFIG = yaml.safe_load(f)

from data.generators.finance_generator import generate_finance_data
from data.generators.hr_generator import generate_employees, generate_monthly_snapshots
from data.generators.ops_generator import generate_ops_data

finance_df = generate_finance_data(n_days=730, seed=42)
emp_df     = generate_employees(n=500, seed=42)
snap_df    = generate_monthly_snapshots(emp_df, n_months=24, seed=42)
ops_df     = generate_ops_data(n_days=730, seed=42)
print('Data loaded.')
```

```python
# Model 1: Finance Anomaly
from ml.models.finance_anomaly import FinanceAnomalyDetector
m1 = FinanceAnomalyDetector(CONFIG)
m1.train(finance_df)
metrics1 = m1.evaluate(finance_df)
print('Finance Anomaly Detector:', metrics1)
```

```python
# Model 2: Revenue Forecaster
from ml.models.revenue_forecast import RevenueForecaster
m2 = RevenueForecaster(CONFIG)
m2.train(finance_df)
metrics2 = m2.evaluate(finance_df)
print('Revenue Forecaster:', metrics2)
```

```python
# Model 3: HR Attrition Classifier
from ml.models.hr_attrition import AttritionClassifier
m3 = AttritionClassifier(CONFIG)
m3.train(emp_df, snap_df)
importance = m3.get_feature_importance()
print('Feature importance:')
print(importance)
importance.plot(kind='barh', title='Attrition Feature Importance', figsize=(8,4))
plt.tight_layout(); plt.show()
```

```python
# Model 5: Ops Anomaly
from ml.models.ops_anomaly import OpsAnomalyDetector
m5 = OpsAnomalyDetector(CONFIG)
m5.train(ops_df)
metrics5 = m5.evaluate(ops_df)
print('Ops Anomaly Detector:', metrics5)
```
