# Scoring Calibration — Business Health Monitor

Documents scoring weight calibration experiments and how domain weights were chosen.

```python
import sys; sys.path.insert(0, '..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Simulate scoring under different weight configurations
configs = [
    {'name': 'Finance-heavy (BP default)', 'f': 0.40, 'hr': 0.30, 'ops': 0.30},
    {'name': 'Equal weights',              'f': 0.33, 'hr': 0.33, 'ops': 0.34},
    {'name': 'Ops-heavy',                  'f': 0.25, 'hr': 0.25, 'ops': 0.50},
    {'name': 'HR-heavy',                   'f': 0.25, 'hr': 0.50, 'ops': 0.25},
]

# Sample domain scores (simulated)
scenarios = [
    {'label': 'All Healthy', 'f_score': 85, 'hr_score': 80, 'ops_score': 82},
    {'label': 'Finance Crisis', 'f_score': 30, 'hr_score': 75, 'ops_score': 78},
    {'label': 'HR Crisis', 'f_score': 78, 'hr_score': 25, 'ops_score': 80},
    {'label': 'Ops Outage', 'f_score': 80, 'hr_score': 75, 'ops_score': 20},
]

results = []
for cfg in configs:
    for sc in scenarios:
        composite = cfg['f'] * sc['f_score'] + cfg['hr'] * sc['hr_score'] + cfg['ops'] * sc['ops_score']
        results.append({'Config': cfg['name'], 'Scenario': sc['label'], 'Composite': round(composite, 1)})

df = pd.DataFrame(results)
pivot = df.pivot(index='Config', columns='Scenario', values='Composite')
print(pivot)
pivot.plot(kind='bar', figsize=(12,5), title='Composite Score by Weight Config & Scenario')
plt.tight_layout(); plt.show()
```

```python
# Conclusion: Finance-heavy (0.40/0.30/0.30) provides best signal separation
# Finance crises produce larger score drops, which is appropriate for a business health system
print('Selected weights: Finance=0.40, HR=0.30, Operations=0.30')
print('Rationale: Finance crises tend to be the most immediately existential for a business.')
```
