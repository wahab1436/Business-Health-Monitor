# Finance EDA — Business Health Monitor

Exploratory analysis of synthetic finance data: distributions, seasonality, and anomaly patterns.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from data.generators.finance_generator import generate_finance_data

df = generate_finance_data(n_days=730, seed=42)
df['date'] = pd.to_datetime(df['date'])
print(df.shape)
df.head()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 8))
df.plot(x='date', y='revenue', ax=axes[0,0], title='Revenue Over Time', color='steelblue')
df.plot(x='date', y='profit_margin', ax=axes[0,1], title='Profit Margin %', color='purple')
df.plot(x='date', y='cash_flow', ax=axes[1,0], title='Cash Flow', color='green')
df[df['anomaly_flag']].plot.scatter(x='date', y='revenue', ax=axes[1,1],
    title='Anomaly Dates', color='red', alpha=0.6)
plt.tight_layout()
plt.show()
print(f'Anomaly rate: {df["anomaly_flag"].mean()*100:.2f}%')
```

```python
df['month'] = df['date'].dt.month
df.groupby('month')['revenue'].mean().plot(kind='bar', title='Avg Revenue by Month',
    figsize=(10,4), color='steelblue')
plt.tight_layout(); plt.show()
```
