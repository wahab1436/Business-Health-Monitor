# HR EDA — Business Health Monitor

Attrition driver analysis, department breakdowns, satisfaction distributions.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
from data.generators.hr_generator import generate_employees, generate_monthly_snapshots

emp_df  = generate_employees(n=500, seed=42)
snap_df = generate_monthly_snapshots(emp_df, n_months=24, seed=42)
print('Employees:', emp_df.shape)
print('Snapshots:', snap_df.shape)
emp_df.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

dept_attr = emp_df.groupby('department')['attrition'].mean() * 100
dept_attr.plot(kind='bar', ax=axes[0], title='Attrition Rate by Department (%)', color='tomato')

dept_sat = emp_df.groupby('department')['satisfaction_score'].mean()
dept_sat.plot(kind='bar', ax=axes[1], title='Avg Satisfaction by Department', color='steelblue')
axes[1].axhline(3.5, color='green', linestyle='--', label='Target 3.5')
axes[1].legend()

plt.tight_layout(); plt.show()
print(f'Overall attrition rate: {emp_df["attrition"].mean()*100:.1f}%')
```

```python
# Satisfaction distribution
emp_df['satisfaction_score'].hist(bins=30, figsize=(10,4), color='purple', alpha=0.7)
plt.title('Employee Satisfaction Score Distribution')
plt.xlabel('Satisfaction (1-5)')
plt.show()
```

```python
print('Attrition correlations:')
print(emp_df[['attrition','satisfaction_score','tenure_months','salary','absenteeism_days']].corr()['attrition'].round(3))
```
